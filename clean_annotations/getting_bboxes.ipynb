```python
import pickle
import numpy as np

clean_annotations = np.load('clean_annotations.npy')

n_unlabeled = 10000

with open('cell_dataset.pkl', 'rb') as f:
    all_data = pickle.load(f)

all_features = all_data['feats']  # (N, D)
annotation_indices = clean_annotations[0]
print("annotation_indices.shape",annotation_indices.shape)
bboxes = all_data["bboxes"]
print("bboxes.shape",len(bboxes))
unlabeled_indices = ~np.isin(np.arange(len(all_data["bboxes"])), annotation_indices)
unlabeled_indices = np.random.choice(len(unlabeled_indices), n_unlabeled, replace=False)


labeled_bboxes = all_data["bboxes"][annotation_indices]
unlabeled_bboxes = all_data["bboxes"][unlabeled_indices]
print("len de unlabeled_bboxes" ,len(unlabeled_bboxes))
print("len de labeled_bboxes" ,len(labeled_bboxes))

np.save(f'unlabeled_bboxes_{n_unlabeled}.npy', unlabeled_bboxes)
```
